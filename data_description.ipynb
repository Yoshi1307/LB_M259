```python
# 2.4
# Min-Max-Skalierung von Hand für die Schüsse
# Wir berechnen das Minimum und Maximum für die Spalte 'HS'
hs_min = df['HS'].min()
hs_max = df['HS'].max()

# Die Formel anwenden
df['HS_Skaliert'] = (df['HS'] - hs_min) / (hs_max - hs_min)

# Das Gleiche für die Auswärtsschüsse 'AS'
df['AS_Skaliert'] = (df['AS'] - df['AS'].min()) / (df['AS'].max() - df['AS'].min())

# Kontrolle der ersten Werte
print(df[['HS', 'HS_Skaliert', 'AS', 'AS_Skaliert']].head(8740))
```
